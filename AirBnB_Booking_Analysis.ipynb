## 3. ***Data Wrangling***

### Data Wrangling Code

```python
#Handling missing values
df = df.dropna(subset=['name','host_name'])   #Dropping missing rows from name and host name.
df.isnull().sum()
```

```python
# dropping column "last_review" as it of no use for us
df = df.drop(columns=['last_review'])
df['reviews_per_month'] = df['reviews_per_month'].replace(np.nan,0)   # replacing null values with 0
df.isnull().sum()
```

```python
df.info()
```

```python
# handling outliers
num_features=['price','minimum_nights','calculated_host_listings_count','availability_365']

plt.figure(figsize=(15,5))
# boxplot of numerical features
sns.boxplot(data=df[num_features])
plt.show()
```

```python
# we are going to replace the datapoints with upper and lower bound of all the outliers

def clip_outliers(df):
    for col in df[num_features]:
        # using IQR method to define range of upper and lower limit.
        q1 = df[col].quantile(0.25)
        q3 = df[col].quantile(0.75)
        iqr = q3 - q1
        lower_bound = q1 - 1.5 * iqr
        upper_bound = q3 + 1.5 * iqr

        # replacing the outliers with upper and lower bound
        df[col] = df[col].clip(lower_bound, upper_bound)
    return df
```

```python
# using the function to treat outliers
df = clip_outliers(df)
```

```python
# checking the boxplot after outlier treatment
plt.figure(figsize=(15,5))
# boxplot of numerical features
sns.boxplot(data=df[num_features])
plt.show()
```

```python
# checking for distribution after treating outliers.
for col in num_features:
  fig, ax =plt.subplots(1,2, figsize=(10,3))
  sns.distplot(df[col], ax=ax[0]).set(title="Distribution")
  sns.boxplot(df[col], ax=ax[1]).set(title="Box Plot")
  plt.suptitle(f'{col.title()}',weight='bold')
  plt.tight_layout()
```

### What all manipulations have you done and insights you found?

- column "last_review" dropped as it is not important for analysis.
- dropped missing values from "name" and 'host_name' as this was less than 5% of the data.
- Replaced missing values in column "reviews_per_month" with 0.
- Removed outliers from numerical columns for better analysis.

#### Chart - 9 - Correlation Heatmap

```python
# Calculate pairwise correlations between columns
cor_features = ['id','host_id','latitude','longitude','price','minimum_nights','number_of_reviews','reviews_per_month','calculated_host_listings_count','availability_365']
corr = df[cor_features].corr()
corr
```

```python
# Set the figure size
plt.figure(figsize=(12,6))

# Visualize correlations as a heatmap
sns.heatmap(corr, cmap='BrBG',annot=True)

# Display heatmap
plt.show()
```

##### 1. Why did you pick the specific chart?

Heat map to show corelation between different features.

##### 2. What is/are the insight(s) found from the chart?

- There is a moderate positive correlation (0.23) between the calculated_host_listings_count column and the availability_365 column, which suggests that hosts with more listings tend to have more days of availability in the next 365 days.
- There is a moderate positive correlation (0.58) between the host_id and id columns, which suggests that hosts with more listings are more likely to have unique host IDs.
- There is a weak positive correlation (0.17) between the price column and the calculated_host_listings_count column, which suggests that hosts with more listings tend to charge higher prices for their listings.
